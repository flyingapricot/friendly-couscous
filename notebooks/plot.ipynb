# Plotting Guide

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

iris = load_iris()
data = pd.DataFrame(iris.data, columns=iris.feature_names)

#Extract required data
sepal_length = data['sepal length (cm)']
petal_length = data['petal length (cm)']


# Peek first 5 rows using .head
sepal_length.head()

# Peek last 5 rows using .tail
sepal_length.tail()

# Plot graph
#First is x-axis, Second is y-axis, label of the curve
plt.plot(sepal_length,petal_length,label="Sepal against Petal")
plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")
plt.title("Sepal against Petal")
plt.show()


#df.loc[condition] => return a slice of df where the condition satisfied .index extracts the index of the slice
val1 = data.loc[data['petal length (cm)'] > 5].index
List1 = data.loc[val1] # Can Be Plotted

```

```python
# Train, Test, Split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(iris['data'], iris['target'], random_state=0)


# create dataframe from data in X_train
# label the columns using the strings in iris_dataset.feature_names
iris_dataframe = pd.DataFrame(X_train, columns=iris.feature_names)
iris_dataframe.head()

# create a scatter matrix from the dataframe, color by y_train
from pandas.plotting import scatter_matrix
grr = pd.plotting.scatter_matrix(iris_dataframe, c=y_train, figsize=(15, 15),
marker='o', hist_kwds={'bins': 20})
```
